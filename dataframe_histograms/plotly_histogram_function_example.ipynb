# Plotly Histogram Function Example

This notebook creates a random pandas dataframe named `df` with 15 variables named `A` through `O`, then defines a function named `create_histograms()` that loops through the dataframe columns and creates Plotly Express histogram charts colored by a target variable.

Example function call:

```python
create_histograms(df, "D")
```


## 1. Import libraries

We will use:

- `pandas` for the dataframe
- `numpy` for random data generation
- `plotly.express` for interactive histograms


````python
import pandas as pd
import numpy as np
import plotly.express as px
````

## 2. Create example dataframe

The dataframe is named `df` and contains 15 variables named with letters of the alphabet from `A` through `O`.

The variables are a mix of numeric and categorical columns.

````python
# Set seed so results are reproducible
np.random.seed(42)

# Number of rows
n = 20

# Create random dataset with 15 variables: A through O
df = pd.DataFrame({
    "A": np.random.randint(18, 70, n),                         # Numeric
    "B": np.random.normal(50000, 12000, n).round(2),          # Numeric
    "C": np.random.choice(["Male", "Female"], n),             # Categorical
    "D": np.random.choice(["Yes", "No"], n),                  # Categorical target example
    "E": np.random.uniform(0, 100, n).round(1),               # Numeric
    "F": np.random.choice(["Red", "Blue", "Green"], n),       # Categorical
    "G": np.random.poisson(5, n),                             # Numeric
    "H": np.random.choice(["Low", "Medium", "High"], n),      # Categorical
    "I": np.random.randint(1, 1000, n),                       # Numeric
    "J": np.random.choice(["A", "B", "C", "D"], n),           # Categorical
    "K": np.random.normal(100, 15, n).round(2),               # Numeric
    "L": np.random.choice(["North", "South", "East", "West"], n), # Categorical
    "M": np.random.randint(0, 2, n),                          # Numeric / Binary
    "N": np.random.choice(["Cat", "Dog", "Bird"], n),         # Categorical
    "O": np.random.uniform(1000, 5000, n).round(2)            # Numeric
})

# Display dataframe
df
````

## 3. Define histogram function

The function:

- Takes `df` as an input
- Takes `TargetVariable` as an input
- Loops over each column in the dataframe
- Uses the current column as the histogram x-axis
- Colors the histogram by `TargetVariable`
- Sets the x-axis label to the current column name
- Shows the legend

The function skips the target variable itself so it does not create a histogram where the x-axis and color variable are the same column.

````python
def create_histograms(df, TargetVariable):
    """
    Create histogram plots for every column in a dataframe.

    Parameters
    ----------
    df : pandas.DataFrame
        Input dataframe.

    TargetVariable : str
        Column name used for histogram coloring.
    """

    # Check that TargetVariable exists in the dataframe
    if TargetVariable not in df.columns:
        raise ValueError(f"{TargetVariable} is not a column in the dataframe.")

    # Optional custom colors for common target values
    # Update/remove as needed
    target_colors = {
        "Yes": "red",
        "No": "blue",
        1: "red",
        0: "blue",
        "Stayed": "blue",
        "Left": "red"
    }

    # Loop through all columns in dataframe
    for col in df.columns:

        # Skip the target variable itself
        if col == TargetVariable:
            continue

        # Create histogram
        fig = px.histogram(
            df,
            x=col,
            color=TargetVariable,
            color_discrete_map=target_colors,
            title=f"{col} Distribution Colored by {TargetVariable}",
            labels={
                col: col,
                TargetVariable: TargetVariable
            },
            barmode="overlay"
        )

        # Show legend and set x-axis title
        fig.update_layout(
            showlegend=True,
            xaxis_title=col
        )

        # Show plot
        fig.show()
````

## 4. Run the function

This will create one histogram for every dataframe column except `D`.

Since `D` contains `Yes` and `No`, each histogram will be colored by those categories.

````python
create_histograms(df, "D")
````

## 5. Expected output

Running:

```python
create_histograms(df, "D")
```

will generate histograms for these columns:

```text
A, B, C, E, F, G, H, I, J, K, L, M, N, O
```

The function skips column `D` because `D` is the target variable used for color.

Each Plotly histogram will include:

- The current dataframe column on the x-axis
- Bars colored by the values in column `D`
- A visible legend
- An interactive Plotly chart with hover labels, zooming, panning, and legend filtering


## Example chart descriptions

For numeric columns such as `A`, `B`, `E`, `G`, `I`, `K`, and `O`, Plotly will create numeric histograms.

For categorical columns such as `C`, `F`, `H`, `J`, `L`, and `N`, Plotly will create count-based bar-style histograms.

Example chart title:

```text
A Distribution Colored by D
```

Example legend:

